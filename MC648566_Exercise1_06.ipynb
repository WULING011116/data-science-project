```python
import pandas as pd
import io

# 1. 读取缺失值数据集
csv_data = """Make,Colour,Odometer (KM),Doors,Price
Toyota,White,150043.0,4.0,$4,000.00
Honda,Red,87899.0,4.0,$5,000.00
Toyota,Blue,,3.0,$7,000.00
BMW,Black,11179.0,5.0,$22,000.00
Nissan,White,213095.0,4.0,
Toyota,Green,,4.0,$4,500.00
Honda,,95767.0,4.0,$7,500.00
Honda,Blue,,4.0,$7,000.00
Toyota,White,60000.0,,$3,000.00
,White,31600.0,4.0,$9,700.00
"""
car_sales_missing = pd.read_csv(io.StringIO(csv_data))

# 2. 填补缺失值（修改点：去除 inplace=True，改用直接重新赋值，消除红色报错）
car_sales_missing["Odometer (KM)"] = car_sales_missing["Odometer (KM)"].fillna(car_sales_missing["Odometer (KM)"].mean())

# 3. 删除仍含有缺失值的行
car_sales_missing.dropna(inplace=True)

# 4. 加载完整 car_sales 数据集进行行列操作与洗牌练习
car_sales = pd.read_csv("car_sales.csv")

# 5. 创建新列
car_sales["Seats"] = 5
fuel_economy = [7.5, 9.2, 5.0, 9.6, 8.7, 4.7, 7.6, 8.7, 3.0, 4.5]
car_sales["Fuel per 100KM"] = fuel_economy
car_sales["Total fuel used (L)"] = car_sales["Odometer (KM)"]/100 * car_sales["Fuel per 100KM"]
car_sales["Number of wheels"] = 4
car_sales["Passed road safety"] = True

# 6. 删除列
car_sales = car_sales.drop("Total fuel used (L)", axis=1)

# 7. 数据洗牌 (Randomize/Shuffle) 与重置索引
car_sales_shuffled = car_sales.sample(frac=1)
car_sales_shuffled = car_sales_shuffled.reset_index(drop=True)

# 8. 使用 apply 和 lambda 将里程数 (KM) 转换为英里 (Miles)
car_sales["Odometer (KM)"] = car_sales["Odometer (KM)"].apply(lambda x: x / 1.6)
print(car_sales.head())
```
